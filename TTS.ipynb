```python
# Step 1: Install
!pip install transformers torchaudio librosa soundfile -q

from transformers import VitsModel, AutoTokenizer
import torch, torchaudio, os
from IPython.display import Audio

# Step 2: Load Akan model (facebook/mms-tts)
model_id = "facebook/mms-tts-aka"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = VitsModel.from_pretrained(model_id)

# Step 3: Your 5 test sentences - CHANGE THESE
# Use simple Akan without too much ɛ/ɔ for first test
sentences = [
    "Akwaaba, me din de Mawunyo",
    "Medaase paa",
    "Ɛte sɛn",
    "Yɛbɛhyia bio",
    "Ghana kasa yɛ fɛ"
]

os.makedirs("tts_output", exist_ok=True)

for i, text in enumerate(sentences):
    try:
        inputs = tokenizer(text, return_tensors="pt")
        with torch.no_grad():
            output = model(**inputs).waveform
        # mms-tts is 16kHz
        path = f"tts_output/akan_sample_{i+1}.wav"
        torchaudio.save(path, output, 16000)
        print(f"Saved {path}: {text}")
        display(Audio(path))
    except Exception as e:
        print(f"Failed on: {text} | Error: {e}")
        # Try fallback - replace special chars
        fallback = text.replace("ɛ","e").replace("ɔ","o").replace("Ɛ","E").replace("ɔ","o")
        inputs = tokenizer(fallback, return_tensors="pt")
        with torch.no_grad():
            output = model(**inputs).waveform
        path = f"tts_output/akan_sample_{i+1}_fallback.wav"
        torchaudio.save(path, output, 16000)
        print(f"Saved fallback {path}")
```
